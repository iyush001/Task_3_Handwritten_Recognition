# Handwritten Character Recognition

## Objective
Recognize handwritten digits from the MNIST dataset using a CNN.


```

import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

# Load MNIST
(X_train, y_train), (X_test, y_test) = tf.keras.datasets.mnist.load_data()

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)

# Show sample images
plt.figure(figsize=(8, 4))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(X_train[i], cmap="gray")
    plt.title(y_train[i])
    plt.axis("off")
plt.show()

# Normalize
X_train = X_train / 255.0
X_test = X_test / 255.0

# Add channel dimension
X_train = X_train.reshape(-1, 28, 28, 1)
X_test = X_test.reshape(-1, 28, 28, 1)

# CNN
model = Sequential([
    Conv2D(32, (3, 3), activation="relu", input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation="relu"),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(128, activation="relu"),
    Dense(10, activation="softmax")
])

model.summary()

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

history = model.fit(
    X_train,
    y_train,
    epochs=5,
    validation_split=0.1
)

loss, accuracy = model.evaluate(X_test, y_test)
print("Test Accuracy:", accuracy)

# Accuracy graph
plt.plot(history.history["accuracy"])
plt.plot(history.history["val_accuracy"])
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend(["Training", "Validation"])
plt.title("Training Accuracy")
plt.show()

# Predictions
predictions = model.predict(X_test)

for i in range(10):
    print("Actual:", y_test[i], "Predicted:", np.argmax(predictions[i]))

# Display predictions
plt.figure(figsize=(10, 5))
for i in range(10):
    predicted = np.argmax(predictions[i])
    plt.subplot(2, 5, i + 1)
    plt.imshow(X_test[i].reshape(28, 28), cmap="gray")
    plt.title("Actual: " + str(y_test[i]) + "\nPredicted: " + str(predicted))
    plt.axis("off")
plt.show()

model.save("handwritten_digit_model.keras")
print("Model saved.")
```

## Conclusion

A CNN was trained on MNIST handwritten digit images. The images were normalized and passed through convolution and pooling layers before classification into digits 0 to 9.
